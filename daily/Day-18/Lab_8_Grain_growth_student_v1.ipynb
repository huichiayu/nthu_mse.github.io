## Part 1. Grain grwoth simulation

The governing equations are 

1) free energy:

$$f(\eta_1,\eta_2, ...,\eta_p) = -\frac{\alpha}{2}\sum_{i=1}^p \eta_i^2 + \frac{\beta}{4}\bigg(\sum_{i=1}^p\eta_i^2\bigg)^2 + \bigg(\gamma-\frac{\beta}{2}\bigg)\sum_{i=1}^p \sum_{j\neq i}^p \eta_i^2 \eta_j^2$$

2) kinetics:

$$\frac{\partial \eta_i}{\partial t} = -L \bigg(\frac{\partial f}{\partial \eta_i} - \kappa^2 \nabla^2 \eta_i \bigg),$$

where

$$\begin{split}\frac{\partial f}{\partial \eta_i} & = -\alpha \eta_i + \frac{\beta}{4}\cdot 2 \cdot \bigg(\sum_{i=1}^p \eta_i^2\bigg) \cdot 2 \eta_i + \bigg( \gamma - \frac{\beta}{2}\bigg) 2 \eta_i \sum_{j\neq i}^p \eta_j^2 \\ & = -\alpha \eta_i + \beta \eta_i \bigg(\sum_{i=1}^p \eta_i^2 - \sum_{j\neq i}^p \eta_i^2 \bigg) + 2 \gamma \eta_i \sum_{j \neq i}^p \eta_j^2 \\ & = -\alpha \eta_i + \beta \eta_i^3 + 2 \gamma \eta_i \sum_{j \neq i}^p \eta_j^2 = -\alpha \eta_i + \beta \eta_i^3 + 2 \gamma \eta_i \bigg(\sum_{i=1}^p \eta_i^2 - \eta_i^2 \bigg).  \end{split}$$
