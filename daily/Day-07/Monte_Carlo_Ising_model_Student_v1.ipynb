**In this coding practice assignment, we will do a famous Monte Carlo example: Ising model for ferromagnetism.**

The entire simulation involves 
* Selecting a random site in the computational domain.
* Calculate the interaction energy between adjacent neighboring sites (nearest neighbors).
* Samping the successful flip.

<div align="left">
<img src="https://i.ibb.co/sJtXBK41/Ising-1.png" width="400">
</div>


The whole Monte Carlo procedures are
1. Set up a 2D grid system. The initial configuration are random spins either +1 or -1.
2. In the Monte Carlo loop,
* Randomly select a position on the lattice and flip the spin.
* At the selected site, compute the energy difference caused by the trial flip:
 
$$\Delta E_i = E_i^{tri} -E_i^{cur},$$ 

where the current energy at that site is $E_i^{cur} = -J \cdot s_i \cdot \sum_{j} s_j -H\cdot s_i.$ The nearest neighbors are the west, east, south, and north neighbors. The trial energy is $E_i^{tri} = -J \cdot (-s_i) \cdot \sum_{j} s_j -H\cdot (-s_i).$ Thus, the energy change is $\Delta E_i = 2J\cdot s_i \sum_j \cdot s_j + H\cdot s_i.$

* If $\Delta E_i \leq 0$, site energy decreases. The flip is accepted. .
* If $\Delta E_i > 0$, site energy increases. The success of flip will be determined by thermal fluctuations. Samlping using Boltzmann distribution: the higher $\Delta E_i$ the lower change of flip. We calculate a weight using $w=\exp(-\Delta E_i/kT)$, and then randomly generate a number $r$. If $r<w$, the flip is accepted. Otherwe, the flip is rejected.
* If the site is at boundaries, impose periodic boundary condition.
* Repeate the procedure above to the next cycle.
