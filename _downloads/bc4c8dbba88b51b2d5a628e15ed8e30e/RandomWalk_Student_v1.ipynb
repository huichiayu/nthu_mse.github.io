#### Part 2. Count the number of walkers

After the walkers jump, we shall count the numbers of walkers at each location.

1. Create a 1D numpy array to store the final postion of walkers.
1. Next, create a set of bins to count how many walkers are in each bin. To do so, we need to set the edges of each bin. The figure below (left panel) shows an example of bins, in which specifically we have 17 bins and the bin width is $h=10$.
2. To set the edges, we first define the west position ($x_w$) and east position ($x_e$) for the binning region.
3. The west edge of the west most bin will be $w_b = x_{w}-0.5\times h$ and the east edge of the east most bin will be $e_b = x_{e} + 0.5\times h$. Thus, the total number of bins will be $n_{bin} = (e_b - w_b)/h$. Note that `n_bin` needs to be an integer.
5. We can use `np.lispace(wb,eb,n_bin_edge)` to create the edges of bins. The total number of bin edges is `n_bin_edge = n_bin + 1`.
7. Before we plot the result, we need to find the center of each bin such that the bars in the barchart are aligned with their centers. The center of a bin is simply the middle between two adjacent bin edges: `center[i] = (edges[i]+edges[i+1])/2`.
8. Once the bin edges are set, we can use `np.histogram()` to count the numbers of walkers in each bin.
9. Finally, use `plt.bar` to make a barchart showing the binned numbers of walkers.
