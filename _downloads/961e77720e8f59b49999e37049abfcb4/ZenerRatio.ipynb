# Mechanical anisotropy

Zener ratio is an indicator of mehcanil anisotopy:

$$A = \frac{2C_{44}}{C_{11}-C_{12}}.$$

General Hooke's law for 2D orthotropic materials:

$$\begin{bmatrix} \sigma_{11} \\ \sigma_{22} \\\sigma_{12} \end{bmatrix} = \begin{bmatrix} C_{11} & C_{12} & 0 \\ C_{12} & C11 & 0 \\ 0 & 0 & C_{44}  \end{bmatrix} \begin{bmatrix} \varepsilon_{11} \\ \varepsilon_{22} \\ 2 \varepsilon_{12} \end{bmatrix} = \frac{E}{\big( 1+ \nu \big) \big( 1-2\nu\big)}\begin{bmatrix} 1-\nu & \nu & 0 \\ \nu & 1-\nu & 0 \\ 0 & 0 & \frac{1-2\nu}{2}  \end{bmatrix} \begin{bmatrix} \varepsilon_{11} \\ \varepsilon_{22} \\ 2 \varepsilon_{12} \end{bmatrix}, $$

where $E$ is Young modulus and $\nu$ is Poisson's ratio.

```python
import numpy as np
import matplotlib.pyplot as plt

def zener_ratio(C11, C12, C44):
    """Zener anisotropy ratio A = 2*C44 / (C11 - C12)."""
    return 2.0 * C44 / (C11 - C12)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def youngs_modulus_cubic(C11, C12, C44, num_points=360):
    """
    Returns theta (0..2pi) and directional Young's modulus E(theta)
    for a cubic crystal in the xy-plane.
    """

    # Compliance matrix components
    S11 = (C11 + C12) / ((C11 - C12) * (C11 + 2*C12))
    S12 = -C12 / ((C11 - C12) * (C11 + 2*C12))
    S44 = 1.0 / C44

    theta = np.linspace(0, 2*np.pi, num_points)
    c = np.cos(theta)
    s = np.sin(theta)

    # cos^2 * sin^2 = (1/4) sin^2(2θ)
    cs2 = c**2 * s**2

    # 1/E(θ) formula
    invE = S11 - 2 * (S11 - S12 - 0.5*S44) * cs2
    E = 1.0 / invE

    return theta, E
```

```python
# Your cases
C11_cases = [157.9*20, 347.6*8, 250*8]
C12_cases = [113.7*20, 121.8*8, 50*8]
C44_cases = [70.5*20,  81.5*8,  100*8]

plt.figure(figsize=(6,6), dpi=120)
ax = plt.subplot(111, projection='polar')

for i, (C11, C12, C44) in enumerate(zip(C11_cases, C12_cases, C44_cases), start=1):
    th, E = youngs_modulus_cubic(C11, C12, C44)
    ax.plot(th, E, label=f"Case {i}")

ax.set_title("Directional Young's Modulus vs Angle\n(Polar Plot)")
ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
plt.tight_layout()
plt.show()

for z in range(3):
    Zr = zener_ratio(C11_cases[z], C12_cases[z], C44_cases[z])
    print("Zener = ", Zr)
```
