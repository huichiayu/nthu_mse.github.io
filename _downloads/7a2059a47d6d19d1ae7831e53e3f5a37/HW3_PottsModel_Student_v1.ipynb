Potts model (named after Renfrey Potts) is a generalized Ising model. Instead of only 2 states (`+1` or `-1`) at each lattice site, Potts model allows an arbitrary state (any integer number) at each site: $s_i$, state value at site $i$. The value of "state" acts as grain ID. In this homework, you will write a Potts model code to simulate the process of grain growth. As in the Ising model, you need to calculate the energy at each site and calculate the energy difference before and after a trial flip. Each trial flip is choosing between the possible states. The energy needs to consider here will be grain boundary energy at each site. For example, at site $i$:

$$E_i = J \sum_{j\neq i} \big(1-\delta_{s_i s_j} \big),$$


where $J$ is the interaction constant, the index $j$ indicates the nearest neighbor sites to site $i$, and $\delta_{s_i s_j}$ is the Kronecker delta. Mathematically, the Krknecker delta is defined as 

$$\delta_{s_i s_j} = 1 ~~\text{if}~~ s_i = s_j$$

$$\delta_{s_i s_j} = 0 ~~\text{if}~~ s_i \neq s_j.$$

The summation indicates that, for site $i$, the grain boundary energy is determined by examining the states at the 8 neighboring sites ($j$). If the states between the neighbor $j$ and site $i$ are different, that contributes to 1 $J$ of energy. Conversely, if the states are the same, no energy contribution to site $i$. Simply speaking, if $j\neq i,$ there is a grain boundary between the two sites. 


The acceptance or rejection of trial flip is similar to that in Ising model. The system will evolve towards reduction of total grain boundary energy, and thus the grain size increases.
