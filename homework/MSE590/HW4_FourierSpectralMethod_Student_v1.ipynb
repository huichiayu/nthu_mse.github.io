\nabla^2 \phi^{(n)} \big],$$

$$ F\big[\phi^{(n+1)}\big] = F\big[\phi^{(n)}\big] - \Delta t M \omega^2 \big\{F \big[ f'(\phi^{(n)}) \big] + \varepsilon^2 \omega^2 F\big[ \phi^{(n)} \big] \big\}.$$
