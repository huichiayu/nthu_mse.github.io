### Part 1. Walk through PCA.

Here, let's do some hands-on PCA practices.  We will break down the process into several small steps.

**A.** We have a data set containing 10 data points. Let's plot those data points and take a look. This is a dataset of 2D data points: 2 features -- $x$ and $y$. (or more mathy $x_1$ and $x_2$)


````python
import numpy as np
import matplotlib.pyplot as plt

# data point
x = np.array([2.5, 0.5, 2.2, 1.9, 3.1, 2.3, 2.0, 1.0, 1.5, 1.1])
y = np.array([2.4, 0.7, 2.9, 2.2, 3.0, 2.7, 1.6, 1.1, 1.6, 0.9])

# plot the data points
plt.scatter(x, y)
plt.axis('equal')
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")
plt.show()
````

---
**B.** Calculate the mean of x and mean of y.

$$\bar{x} = \frac{1}{N} \sum_{i=1}^N x_i~~~\text{and}~~~\bar{y} = \frac{1}{N} \sum_{i=1}^N y_i.$$


````python
x_mean = sum(x)/len(x)
y_mean = sum(y)/len(y)

print(x_mean)
print(y_mean)
````

---
**C.** Now, we need to calculate the covariance matrix of the new data set: 

$$CovM = \left[ \begin{array}{cc} cov(X,X) & cov(X,Y) \\ cov(Y,X) & cov(Y,Y) \end{array} \right],$$

where the covariance between two sets of values can be obtained using the formula below. For example, $cov(X,Y)$ is 

$$ cov(X,Y) = \frac{\sum_{i=1}^N (X_i-\bar{X}) (Y_i - \bar{Y})}{N-1},$$

where $n$ is the number of values in the set, and the overhead bar indicates the mean value.

**Calculate the covariance matrix in the cell below.**

````python
c_xx = sum((x - x_mean)*(x - x_mean))/(len(x)-1)
c_yy = sum((y - y_mean)*(y - y_mean))/(len(y)-1)
c_xy = sum((x - x_mean)*(y - y_mean))/(len(y)-1)

CovM = np.array([[c_xx, c_xy],[c_xy, c_yy]])

print(CovM)
````

**A covariance matrix answers:**

>If one variable increases, does another tend to increase, decrease, or stay unrelated?
>
>* diagonal entries → variances
>* off-diagonal entries → relationships between features

&#9989; Do This - Use your own words, explain why the off-diagonal entries indicate the relationship between features.

---
**D.** Calculate the eigenvalues and eigenvectors of the covariance matrix. You can use existing function to do this. For example, `np.linalg.eig` or `scipy.linalg.eig` in SciPy library. 

````python
e_vals, e_vecs = np.linalg.eig(CovM)

print("Eigenvalues:", e_vals)
print("Eigenvectors:\n", e_vecs)
````

Let's plot these eigenvector with the data points. A line on a 2D plane is defined by $y = wx +b$, where $w$ and $b$ are the slope and intercept, respectively. Assuming the intercept is zero, the line 
along eigenvector $u_1$ is  

$$y = wx + 0 = \bigg(\frac{u_y}{u_x}\bigg) x.$$




The cell below plots the axes along the 2 eigenvectors.

&#9989; Do This - Describe what you observed between the relationships between the data points and the two eigenvectors.

````python
# eigenvector 1
ev1 = e_vecs[0,:]
px1 = np.array([-1,5])
py1 = ev1[1]/ev1[0]*px1

# eigenvector 2
ev2 = e_vecs[1,:]
px2 = np.array([-1,0.9])
py2 = ev2[1]/ev2[0]*px2

print(px1,py1)
print(px2,py2)

# plotting
plt.clf()

plt.arrow(np.max(px1), np.max(py1), np.min(px1)-np.max(px1), np.min(py1)-np.max(py1),
          head_width=0.1,
          head_length=0.2,
          length_includes_head=True,
          color='b')

plt.arrow(np.min(px2), np.max(py2), np.max(px2)-np.min(px2), np.min(py2)-np.max(py2),
          head_width=0.1,
          head_length=0.2,
          length_includes_head=True,
          color='r')

plt.grid(True)
plt.gca().set_aspect('equal')
plt.xlim(-1.01, 4)
plt.ylim(-1.01, 4)
plt.xlabel("x")
plt.ylabel("y")

plt.scatter(x, y)
````

---
**E.** Transfer the original data using the two eigenvectors.  

The cell below calculates the values of the data points projected on the blue and red axes. 

````python
nA = np.zeros((2,len(x)))
nA[0,:] = x
nA[1,:] = y

# obtain the data in the new coordinate from PCA
n_x = np.matmul(ev1,nA)
n_y = np.matmul(ev2,nA)

# plot the result
plt.scatter(n_x,n_y)
plt.grid(True)
plt.axis('equal')
plt.show()
````

**Recall what eigenvalues and eigenvectors of a matrix are.** You should have learned eigenpairs in your Engineering Math class years ago. 

>$$\mathbf{A} \mathbf{v} = \lambda \mathbf{v},$$ 
>
>where $\mathbf{A}$ is a square matrix, $\mathbf{v}$ is a column vector, and $\lambda$ is the eigenvalue of the matrix. $\lambda$ is obtained by solving
>
>$$\text{det}\bigg( \mathbf{A} - \lambda \mathbf{I} \bigg) = 0,$$
>
>where $\mathbf{I}$ is an identity matrix. There will be N number of eigenvalues of a NxN matrix. Each eigenvalue is associated with an eigenvector.

A matrix multiply with a vector is essentially a **"rotation"** (transformation) of that vector in a high dimensional space. Eigenvectors are in fact the direction of the coordinate axes in that space. All eigenvectors are orthogonal to each other (inner product is zero), as the coordinate axes are orthogonal. 

So, the eigenvectors are the coordinate axes of the space that covariance matrix can rotate. If we observe the data along the basis axes, the data points are stretched as far as possible. The eigenvector associated with a larger eigenvalue is along the axis that stretches the data points more. **If we set number of PCA components to 10, the PCA will pick up the first 10 eigenvectors (axes), along which the data points are separated most.**

## Part 2. Image regression.

In the previous class, we built an ANN regressor to predict the temperatures of Ising-model simulated images. Here we will do the same for predicting the total free energy of a spinodal decomposed bi-continuous structure. 

$$\mathcal{F} = \int f(\phi) + \frac{1}{2} \varepsilon^2 \big(\nabla \phi \big)^2 d\Omega.$$

Below, we will use a dataset from https://mds-book.org/Content/datasets. This dataset contains 17866 Cahn-Hilliard phase-field simulation results.

**Following the same procedure as in the preivous class.** We'll

* Load the image data.
* Extract some images for building the ANN model.
* Examine the image data. 

````python
import matplotlib.pyplot as plt
from mdsdata import load_CahnHilliard

images, energies = load_CahnHilliard()

# or: 
# from mdsdata import MDS3
# images, energies = MDS3.load_data(return_X_y=True)

print("The dataset contains", images.shape[0], "images.")
print("They are", images.shape[1], "x", images.shape[2], "pixel in size.")
print(f"The minimum energy is: {energies.min(): .1f}")
print(f"The maximum energy is: {energies.max(): .1f}")
````

````python
# extract 2000 sets out of the 17866 randomly
idx = np.random.choice(17866, 2000, replace=False)   # 1000 unique indices
subset_im = images[idx] 
subset_lb = energies[idx]

n_samples, h, w = subset_im.shape

print(n_samples, h, w)
````

````python
#Cool slider to browse all of the images.
from ipywidgets import interact
def browse_images_I(images, labels):
    n = len(images)
    def view_image(i):
        plt.imshow(images[i], cmap='viridis', interpolation='nearest')
        plt.title('%s' % f"E={labels[i]:.2f}")
        plt.axis('off')
        plt.show()
    interact(view_image, i=(0,n-1))
````

````python
browse_images_I(subset_im, subset_lb)
````

* Load the libraries for building ANN model.
* Reshape the 2D images to 1D row feature vectors.

````python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_absolute_error
````

````python
# flatten each image using reshape
# X_np has 2000 rows, each row is a 1x(64x64) row vector
# Note that X_np and y_np are in the format of np array
# y: output is the energy
X_np = subset_im.reshape(subset_im.shape[0],-1)
y_np = subset_lb
````

````python
print("X_np shape:", X_np.shape)
print("y_np shape for sklearn:", y_np.shape)

# Make sure y is 2D: (N, 1)
if y_np.ndim == 1:
    y_pt = y_np.reshape(-1, 1)
print("y_pt shape for PyTorch:", y_pt.shape)
````

* Train-test split.
* Scaling.
* Data conversion: from numpy to PyTorch tensor.
* Batch split for training.

````python
from sklearn.model_selection import train_test_split

# Train/test split
X_train_np, X_test_np, y_train_np, y_test_np = train_test_split(
    X_np, y_pt, test_size=0.3, random_state=42)

y_scaler = StandardScaler()
y_train_np_scaled = y_scaler.fit_transform(y_train_np)  # (N_train, 1)
y_test_np_scaled  = y_scaler.transform(y_test_np)       # (N_test, 1)

# To torch tensors (X is already roughly scaled: 0/1; we just cast to float)
X_train_torch = torch.from_numpy(X_train_np).float()
X_test_torch  = torch.from_numpy(X_test_np).float()
y_train_torch = torch.from_numpy(y_train_np_scaled).float()
y_test_torch  = torch.from_numpy(y_test_np_scaled).float()

print("X_train_torch:", X_train_torch.shape)
print("y_train_torch:", y_train_torch.shape)

batch_size = 32

train_dataset = TensorDataset(X_train_torch, y_train_torch)
test_dataset  = TensorDataset(X_test_torch,  y_test_torch)

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
test_loader  = DataLoader(test_dataset,  batch_size=batch_size, shuffle=False)
````

&#9989; Do This - Use your own words to explain the benefits of batch training.

---

* Initialize ANN model.
* Set up backward correction optimizer.

````python
class ANNRegressor(nn.Module):
    def __init__(self, in_features=1024, hidden1=512, hidden2=256):
        super(ANNRegressor, self).__init__()
        self.net = nn.Sequential(
            nn.Linear(in_features, hidden1),
            nn.ReLU(),
            nn.Linear(hidden1, hidden2),
            nn.ReLU(),
            nn.Linear(hidden2, 1)   # 1 output (scaled temperature)
        )

    def forward(self, x):
        return self.net(x)
````

````python
model = ANNRegressor(4096, 2048, 512)
print(model)

# If using GPU:
# model = ANNRegressor().to(device)

criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=1e-3)
````

* Train the model.

````python
epochs = 100
tMSE = []
n_epo = []

import time
start = time.time()

# training step
for epoch in range(1, epochs + 1):
    model.train()
    running_loss = 0.0

    # train at batch size
    for X_batch, y_batch in train_loader:
        # If using GPU:
        # X_batch = X_batch.to(device)
        # y_batch = y_batch.to(device)

        # forward propagtion
        optimizer.zero_grad()
        y_pred = model(X_batch)
        loss = criterion(y_pred, y_batch)
        
        # backward correction
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * X_batch.size(0)

    # check loss function value
    epoch_loss = running_loss / len(train_dataset)
    tMSE.append(epoch_loss)
    n_epo.append(epoch)
    
    if epoch % 10 == 0 or epoch == 1:
        print(f"Epoch {epoch:3d}/{epochs}, Train MSE: {epoch_loss:.6f}")

stop = time.time()
elapse = stop - start

print(f"training time: {elapse:3f}")
````

* Plot the training error versus epoch.

````python
plt.plot(n_epo, tMSE)
plt.grid(True)
plt.xlabel("epoch step")
plt.ylabel("train MSE")
plt.xlim(0, 100)
plt.ylim(-1, 30)
````

* Evaluate the model using test data.

````python
# run prediction
model.eval()

with torch.no_grad():
    y_test_pred_scaled = model(X_test_torch)  # predictions in scaled space

# convert the PyTorch output to numpy
y_test_pred_scaled_np = y_test_pred_scaled.detach().cpu().numpy()

# invert scaling to get real temperature values
y_test_pred_np = y_scaler.inverse_transform(y_test_pred_scaled_np)
y_test_np_orig = y_test_np  # original temperatures from before scaling

# metrics
r2  = r2_score(y_test_np_orig, y_test_pred_np)
mae = mean_absolute_error(y_test_np_orig, y_test_pred_np)

print(f"R²:  {r2:.4f}")
print(f"MAE: {mae:.6f}")

print("\nFirst 5 samples:")
for i in range(5):
    print(f"sample {i}: pred = {y_test_pred_np[i,0]:.3f}, true = {y_test_np_orig[i,0]:.3f}")
````

````python
plt.scatter(y_test_np_orig,y_test_pred_np)
plt.grid(True)
plt.xlabel("true energy")
plt.ylabel("predicted energy")
````

## Part 3. PCA of the image data.

**You may observe that the image regressor works pretty well for this dataset.** However, each image data point contains 4096 feature values. The computing could be intense. We can use PCA to grab only several most important features to make a new input dataset by projecting the original images to the those **"important axes"**. For example, if we pick 10 PCA components, after projecting the 4096 feature values of an image to the 10 orthogonal axes (eigenvectors), we will have only 10 scalar values (each along one axis).
>Note that one image is a row vector of 4096 freatures. Each eigenvector is also a row vector of 4096 values. A row vector projected on an eigenvector is the inner product of these two vectors, which is a single scalar value. Thus, after transforming an image (4096-featured vector) to the first 10 eigenvectors (PCA components), we obtain 10 scalar numbers, which forms a new feature vector with reduced features, i.e., **reducing the features from 4096 to 10**.

We can expect that **feature reduction** and using more resolved data can significantly decrease the computational burdens.

The cell below use the built-in `PCA` function of `sklearn` library to do PCA. Let's try 64 PCAs.

````python
from sklearn.decomposition import PCA

# number of PCA components
n_components = 64

print("Extracting the top %d eigenimages from %d images" %
      (n_components, X_train_np.shape[0]))

# set up PCA
pca = PCA(n_components=n_components,
    svd_solver="randomized",   # replacement for RandomizedPCA
    whiten=False,              # set True if needed
    random_state=42)

# calculate PCA (eigenvector)
pca.fit(X_train_np)

# transform the feature vectors along eigenvectors
X_train_pca = pca.transform(X_train_np)
X_test_pca  = pca.transform(X_test_np)

print("PCA shape:", X_train_pca.shape)
````

**The cells below repeat the tasks in Part 2 to run an ANN model, but with data transformed to the PCA components.**

````python
X_train_torch_pca = torch.from_numpy(X_train_pca).float()
X_test_torch_pca  = torch.from_numpy(X_test_pca).float()

print("X_train_torch:", X_train_torch_pca.shape)
print("y_train_torch:", y_train_torch.shape)

batch_size = 32

train_dataset = TensorDataset(X_train_torch_pca, y_train_torch)
test_dataset  = TensorDataset(X_test_torch_pca,  y_test_torch)

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
test_loader  = DataLoader(test_dataset,  batch_size=batch_size, shuffle=False)
````

````python
model_pca = ANNRegressor(n_components, 2048, 512)
print(model_pca)

criterion = nn.MSELoss()
optimizer = optim.Adam(model_pca.parameters(), lr=1e-3)
````

````python
epochs = 100
tMSE = []
n_epo = []

start = time.time()

# training step
for epoch in range(1, epochs + 1):
    model_pca.train()
    running_loss = 0.0

    # train at batch size
    for X_batch, y_batch in train_loader:
        # If using GPU:
        # X_batch = X_batch.to(device)
        # y_batch = y_batch.to(device)

        # forward propagtion
        optimizer.zero_grad()
        y_pred = model_pca(X_batch)
        loss = criterion(y_pred, y_batch)
        
        # backward correction
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * X_batch.size(0)

    # check loss function value
    epoch_loss = running_loss / len(train_dataset)
    tMSE.append(epoch_loss)
    n_epo.append(epoch)
    
    if epoch % 10 == 0 or epoch == 1:
        print(f"Epoch {epoch:3d}/{epochs}, Train MSE: {epoch_loss:.6f}")


stop = time.time()
elapse = stop - start

print(f"training time: {elapse:2f} s")
````

## Part 4 Eigenimage

As mentioned eariler, eigenvectors are the coordinate axes in a hyper dimension. Every vector in that hyper dimension is a linear combination of those basis coordinate vectors. Each eigenvector has the same size of the original data. The analogy is, for example, in 2D, the coordinate axes, $\vec{u} = 1 \vec{x} + 0 \vec{y}$ and $\vec{v} = 0 \vec{x} + 1 \vec{y}$, are still 2D vectors -- same as the dimensions of the data points.

Thus, we can reshape the eigenvectors to 2D arrays. This 2D arrays are "eigenimages". **Every image in the dataset is in fact a linear combination of these basis images.**

In the cell below, let's plot some eigenimages. Note that the PCA components (eigenvectors) are still 4096 dimensional data. 

````python
print(pca.components_.shape)

eigenimages = pca.components_.reshape((n_components, h, w))
print(h, w)
````

**Plot the eigenimages.**

````python
def plot_gallery(images, titles, h, w, n_row=2, n_col=5):
    """Helper function to plot a gallery of portraits"""
    plt.figure(figsize=(1.7 * n_col, 2.3 * n_row))
    plt.subplots_adjust(bottom=0, left=.01, right=.99, top=.90, hspace=.35)
    for i in range(n_row * n_col):
        plt.subplot(n_row, n_col, i + 1)
        plt.imshow(images[i].reshape((h, w)), cmap='viridis')
        plt.title(titles[i], size=12)
        plt.xticks(())
        plt.yticks(())
        
    plt.show()
````

## Part 5 Hyperparameter tuning

We observed that the first several principal components (PCs) can largely decrease the computational burdens of the NN model. **However, how many PCs are good enough?** Like the case of grid search in the SVM coding practice, we should try different numbers of PCs to see the accuracy changes. Let's try varying `n_components` from 10 to 1280 and examine the change in accuracy and training times.

````python
epochs = 100
nMSE = []
walltime = []
n_pc = []

# loop n_components over 10 through 1280 
for pw in range(8):
    n_components = (2**pw)*10

    print(f"n_PCs --> {n_components}")
    
    # repeat the tasks in the cell earlier
    # set up PCA
    pca = PCA(n_components=n_components,
        svd_solver="randomized",   # replacement for RandomizedPCA
        whiten=False,              # set True if needed
        random_state=42)
    
    # calculate PCA (eigenvector)
    pca.fit(X_train_np)

    # transform the vectors along eigenvectors
    X_train_pca = pca.transform(X_train_np)
    X_test_pca  = pca.transform(X_test_np)

    # convert np array to torch tensor
    X_train_torch_pca = torch.from_numpy(X_train_pca).float()
    X_test_torch_pca  = torch.from_numpy(X_test_pca).float()  

    # batch split
    batch_size = 32
    
    train_dataset = TensorDataset(X_train_torch_pca, y_train_torch)
    test_dataset  = TensorDataset(X_test_torch_pca,  y_test_torch)
    
    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
    test_loader  = DataLoader(test_dataset,  batch_size=batch_size, shuffle=False)

    # set up ANN model
    model_pca = ANNRegressor(n_components, 2048, 512)
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model_pca.parameters(), lr=1e-3)

    # train the model
    start = time.time()
    
    # training step
    for epoch in range(1, epochs + 1):
        model_pca.train()
        running_loss = 0.0
    
        # train at batch size
        for X_batch, y_batch in train_loader:
            # If using GPU:
            # X_batch = X_batch.to(device)
            # y_batch = y_batch.to(device)
    
            # forward propagtion
            optimizer.zero_grad()
            y_pred = model_pca(X_batch)
            loss = criterion(y_pred, y_batch)
            
            # backward correction
            loss.backward()
            optimizer.step()
    
            running_loss += loss.item() * X_batch.size(0)
    
        # check loss function value
        epoch_loss = running_loss / len(train_dataset)

    stop = time.time()
    elapse = stop - start  # training time
    walltime.append(elapse)
    
    #  evaluation
    # run prediction
    model_pca.eval()
    
    with torch.no_grad():
        y_test_pred_scaled = model_pca(X_test_torch_pca)  # predictions in scaled space
    
    # convert the PyTorch output to numpy
    y_test_pred_scaled_np = y_test_pred_scaled.detach().cpu().numpy()
    
    # invert scaling to get real temperature values
    y_test_pred_np = y_scaler.inverse_transform(y_test_pred_scaled_np)
    y_test_np_orig = y_test_np  # original temperatures from before scaling
    
    # metrics
    mae = mean_absolute_error(y_test_np_orig, y_test_pred_np)

    print(f"n_PCs: {n_components}, Epoch: {epoch:3d}, training time: {elapse:2f}, MAE: {mae:3f}")
    
    nMSE.append(mae)
    n_pc.append(n_components)
  
````

````python
plt.plot(n_pc,nMSE)
plt.grid(True)
plt.xlabel("n_pca")
plt.ylabel("MAE")
````

````python
plt.plot(n_pc, walltime)
plt.grid(True)
plt.xlabel("n_pca")
plt.ylabel("trainig time")
````
